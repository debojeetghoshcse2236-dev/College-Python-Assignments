Calculating Factorial of 7 with a Loop.

```python
n = 7
factorial = 1

for i in range(1, n + 1):
    factorial = factorial * i

print("Factorial =", factorial)
```

3. Write a Python program to take a number as input and check whether it is a prime number using a for loop.

```python
n = int(input("Enter a number: "))

if n <= 1:
    print("Not a prime number")
else:
    prime = True

    for i in range(2, n):
        if n % i == 0:
            prime = False
            break

    if prime:
        print("Prime number")
    else:
        print("Not a prime number")
```

1. Write a Python program to take an integer as input and find the sum of its digits using a while loop.


```python
n = int(input("Enter a number: "))
sum = 0

while n > 0:
    digit = n % 10
    sum = sum + digit
    n = n // 10

print("Sum of digits =", sum)
```

2. Write a Python program to take an integer n as input and reverse the number using a while loop.
   Example:
   Input: 12345
   Output: 54321


```python
n = int(input("Enter a number: "))
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

print("Reversed number =", reverse)
```

3. Write a Python program to repeatedly ask the user to enter a number until the user enters 0. At the end, print the sum of all the numbers entered (excluding 0).

```python
sum = 0

while True:
    n = int(input("Enter a number: "))

    if n == 0:
        break

    sum = sum + n

print("Sum =", sum)
```

2. Write a Python program to remove all duplicate elements from a list without using set().
   Example:
   Input:  [10, 20, 10, 30, 20, 40]
   Output: [10, 20, 30, 40]

```python
numbers = [10, 20, 10, 30, 20, 40]
unique = []

for n in numbers:
    if n not in unique:
        unique.append(n)

print("List after removing duplicates:", unique)
```

3. Write a Python program to take a list of numbers and create two separate lists:
   - One containing all even numbers
   - One containing all odd numbers

   Example:
   Input:  [12, 7, 5, 18, 20, 9]
   Output:
   Even: [12, 18, 20]
   Odd: [7, 5, 9]

```python
numbers = [12, 7, 5, 18, 20, 9]

even = []
odd = []

for n in numbers:
    if n % 2 == 0:
        even.append(n)
    else:
        odd.append(n)

print("Even:", even)
print("Odd:", odd)
```
